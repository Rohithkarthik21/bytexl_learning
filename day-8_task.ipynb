```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("riturajsingh99/netflix-userbase")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import numpy as np
import os
dataset_path = os.path.join(path, 'Netflix Userbase.csv')

df = pd.read_csv(dataset_path)
df
```

```python
display(df.head())
display(df.tail())
```

```python
print(df.shape)
df.info()
```

```python
print(df.dtypes)
print(df.isna().sum())
```

```python
df.rename(columns={
    "User ID": "user_id",
    "Age": "age",
    "Subscription Type": "sub_type",
    "Monthly Revenue": "mon_revenue"
},inplace=True)
print(df.columns)
```

```python
mean_val=df["mon_revenue"].mean()
print(mean_val)

med_val=df["mon_revenue"].median()
print(med_val)

mode_val=df["mon_revenue"].mode()
print(mode_val)
```
